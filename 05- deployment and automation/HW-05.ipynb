## Question 1: [Exploratory] Research Idea Number One

### Local-language mine-disruption signals in Peruvian copper

#### 1. Market and scope

**Universe:** companies with material exposure to Peruvian copper production, traded in Peru (Bolsa de Valores de Lima, BVL) and/or as US ADRs:

- Southern Copper (SCCO; Toquepala, Cuajone, Tia Maria)
- Buenaventura (BVN; polymetallic, Peruvian assets)
- Freeport-McMoRan (FCX; Cerro Verde)
- MMG (1208.HK; Las Bambas)
- Diversified majors for which Peru is material: BHP, Glencore and Teck (Antamina), Anglo American (Quellaveco)

**Benchmarks:** COPX (copper miners ETF), LME/COMEX copper, a basket of unaffected copper miners, and the S&P/BVL Peru General index.

**Horizon:** event-driven, from 1 day to 3 months. The production and guidance impact is felt over 1 to 4 quarters, but the equity reaction usually starts around the first credible local report.

#### 2. Investment thesis

Peruvian copper supply is unusually exposed to **non-financial operational shocks**: community protests, road blockades, labor strikes, permitting and environmental delays, and water-license disputes. These events are reported first in Spanish by local institutions and local media. English-language financial media and sell-side consensus often pick them up late, partially, or with the wrong severity.

If a disruption removes tonnage or raises unit costs, the affected miner's cash flow falls. Because asset-level models are updated with a lag, the local share or ADR may underreact and then drift as MINEM production data and company guidance confirm the loss. The resolution of a conflict is the mirror-image positive catalyst.

The thesis is **relative, not a copper-price bet**: after controlling for copper, the affected miner should underperform a basket of unaffected copper miners when a severe disruption is under-priced, and outperform when it is resolved.

**Key assumption:** a credible local report can be mapped to lost tonnes within a few days, and the market does not fully price the loss until MINEM data or company guidance confirms it.

#### 3. Potential edge

- **Local language and local geography.** Knowing which road, port, power line or water source is blocked matters. A routine protest far from the mine is not the same as a blockade of the access road or the concentrate port.
- **Fragmented sources.** Mine-level operational information is spread across MINEM, Osinergmin, SENACE, the Defensoria del Pueblo, regional newspapers and community radio. It is not systematically parsed by global quant funds.
- **Thin coverage of asset-level operations.** Diversified miners disclose asset detail with a lag, and pure-play Peruvian miners have limited analyst coverage.
- **Severity mapping.** The edge is not sentiment; it is translating an event into expected lost tonnes and timing.

#### 4. Data and sources

**Checked (public documentation; not yet ingested):**

- MINEM *Boletin Estadistico Minero*: monthly production by company and region.
- Osinergmin supervision and penalty reports.
- OEFA environmental enforcement reports (incidents and sanctions).
- SENACE environmental certification records (Peru; note that SEIA is the equivalent system in Chile).
- Defensoria del Pueblo *Reporte de Conflictos Sociales*: monthly inventory of social conflicts, including mining-related disputes.
- Company filings via SMV (Peru) and SEC (6-K/20-F), plus earnings calls.
- BVL prices, BCRP FX, LME/COMEX copper prices.

**Satellite and port data:**

- Sentinel-2 (free, 10 m) and Sentinel-1 SAR for stockpile, road and port activity; Planet (commercial) for higher cadence.
- AIS data (e.g., MarineTraffic) for concentrate shipments.

**Still to verify:** exact historical archive depth and machine-readability of Defensoria/MINEM/SENACE; satellite revisit frequency versus cloud cover; AIS licensing cost; whether mine-level production is consistently reported at the frequency needed.

#### 5. Research approach

- **Event database:** date, mine, event type (strike, blockade, permit, accident), severity, expected lost tonnes, source and first-report timestamp.
- **Signal:** `disruption_score = expected lost tonnes / annual production`, decayed over 20 trading days; or a binary "high-severity event in the last 20 days".
- **Decision:** long or short the affected miner against an unaffected copper-miner basket or COPX, isolating operational risk from the copper price. Test a separate "resolution" signal for the reversal.
- **Tests:** event study of cumulative abnormal returns around the first Spanish report versus the first English report; panel regression of monthly production surprises on the score; predictive regression of next-month relative return on the score, controlling for copper, FX and the local market.

The approach fits because the information is local and operational, and the trade is a relative-value expression of a supply shock rather than a directional commodity call.

#### 6. Validation and risks

- **Benchmarks:** COPX, copper price, peer-miner basket, S&P/BVL Peru General; buy-and-hold.
- **Reject if:** there is no incremental predictive power after controlling for copper and peer returns; the signal only captures already-public English news; the effect disappears after realistic costs; or event dates cannot be known ex ante.
- **Risks:** low BVL/ADR liquidity; shorting and borrow costs; transaction costs and slippage; ADR-local price divergence; confounding by copper, USD and Peruvian politics; small event sample; false or biased local reports; data licensing; publication delays.
- **First two-week test:** hand-code 30 historical events, measure cumulative abnormal return from the first local report to +20 trading days, and compare with the peer basket. Kill the idea if there is no drift after the first English headline.

## Question 2: [Exploratory] Research Idea Number Two

### ADR-local basis mean reversion in Brazil

#### 1. Market and scope

**Universe:** Brazilian companies with both a B3 listing and a US ADR:

- Petrobras (PBR/PBR.A vs PETR4/PETR3)
- Vale (VALE vs VALE3)
- Itau Unibanco (ITUB vs ITUB4)
- Ambev (ABEV vs ABEV3)
- Gerdau (GGB vs GGBR4)
- Cemig (CIG vs CMIG4)
- Copel (ELP vs CPLE6)

Focus on the 4 to 6 most liquid pairs. **Horizon:** intraday to 10 trading days. Basis deviations are usually short-lived, so this is a relative-value strategy rather than a fundamental holding.

#### 2. Investment thesis

An ADR and the local share are claims on the same cash flows, but they trade in different time zones, currencies and investor bases. Arbitrage should pin the basis, but **limits to arbitrage** allow temporary deviations: capital controls, dividend withholding taxes, FX hedging costs, ADR creation/redemption frictions, and local versus foreign flows.

The basis tends to widen around political or regulatory news, large FX moves, index rebalances, dividend events and US risk-off episodes. It mean-reverts when arbitrageurs or authorized participants act.

The mechanism is **relative value**: long the cheap leg and short the expensive leg. The thesis does not require a view on Brazilian equities or on the real.

**Key assumption:** after adjusting for taxes, borrow costs and corporate actions, the residual basis is a liquidity/flow dislocation rather than fair compensation for risk.

#### 3. Potential edge

- **Local market microstructure and tax knowledge:** ADR ratios, dividend withholding, IOF, PTAX, B3 auction mechanics, short-sale restrictions and borrow availability.
- **Structural blind spot:** many global quant funds treat the ADR and the local line as one asset and ignore the basis. Local brokers and arbitrage desks have an operational advantage.
- **Conditioning information:** combining the basis with foreign-flow and borrow-cost data helps distinguish a temporary dislocation from a persistent repricing.

#### 4. Data and sources

**Checked (public documentation; not yet ingested):**

- B3 and NYSE/Nasdaq prices; B3 USD/BRL futures and BCB PTAX.
- SEC 6-K/20-F filings and depositary-bank notices (e.g., BNY Mellon, Citi, JPMorgan) for ADR ratios and corporate actions.
- B3 foreign-investor flow reports and company dividend calendars.

**Still to verify:** historical intraday data licensing; borrow fees and availability; exact ADR ratios and any changes; withholding-tax treatment; whether foreign-flow data is timely enough for a daily signal.

#### 5. Research approach

- **Basis:** `basis_t = (P_adr_t * ratio * FX_t) / P_local_t - 1`, using synchronous timestamps or lagged prices to avoid stale-price bias.
- **Signal:** z-score of the basis versus a rolling 60-day mean, excluding corporate-action windows and FX jumps. Enter when `abs(z) > 2` and liquidity/borrow conditions are met; exit at `z ~ 0` or after N days.
- **Extensions:** a flow-conditioned signal (basis divergence plus foreign outflow/inflow) and an event filter (dividends, index rebalance).
- **Decision:** long the cheap leg, short the expensive leg, hedge FX; size by liquidity and borrow.
- **Tests:** cointegration and half-life; out-of-sample Sharpe; turnover and maximum drawdown; filtered versus unfiltered strategy.

The approach fits because the signal is directly tradeable, measurable at high frequency, and grounded in a specific market mechanism.

#### 6. Validation and risks

- **Benchmarks:** buy-and-hold each leg; a naive basis strategy; random entry; a peer-pair portfolio.
- **Reject if:** deviations are explained by corporate actions, FX or borrow costs; there is no profit after realistic costs; profits are concentrated in a few episodes; or the strategy breaks down out of sample.
- **Risks:** shorting costs and recall; ADR ratio changes; dividend withholding; FX hedging; capital controls; liquidity gaps; stale prices; structural breaks such as ADR delisting; taxes; and the possibility that the basis is compensation for risk rather than mispricing.
- **First two-week test:** build the daily basis for PBR/PETR4 and VALE/VALE3 over five years, estimate half-life, and compute net P&L after 20 bps round-trip costs plus borrow.

## Question 3: [Exploratory] Research Idea Number Three

### Gaming engagement as a demand nowcast for publishers

#### 1. Market and scope

**Universe:** global video-game publishers with PC/console live-service exposure:

- Electronic Arts (EA)
- Take-Two Interactive (TTWO)
- Roblox (RBLX)
- Ubisoft (UBI.PA)
- CD Projekt (CDR.WA)
- Nexon (3659.T)
- Krafton (259960.KS)
- Embracer (EMBRAC-B.ST)

**Horizon:** 1 to 3 months, around quarterly earnings and major releases. **Benchmarks:** GAMR/ESPO (gaming ETFs), a publisher peer basket, and consensus revenue.

Focus on games where engagement is observable: Steam concurrent players, Twitch hours watched and app-store rankings.

#### 2. Investment thesis

For live-service games, revenue is a function of **engagement** (DAU/MAU, hours, recurring purchases) and **monetization**. Engagement data leads reported bookings and revenue by weeks.

The market anchors on release hype and consensus. Engagement decay and monetization changes (battle pass, price increases, discounts) drive revenue per user. A persistent engagement surprise should predict revenue surprises and guidance revisions.

The macro trend is digital distribution and live-service models, which increase the share of recurring revenue and make engagement more measurable. The risk is that the market over-extrapolates launch spikes and underweights decay.

**Key assumption:** game-level engagement has a stable, measurable relationship with publisher revenue over the next quarter, after controlling for seasonality, portfolio mix and monetization.

#### 3. Potential edge

- **Messy alternative data.** Engagement data is noisy and not standardized. Mapping game-level engagement to company revenue requires portfolio mix, platform splits, monetization and seasonality.
- **Beyond sentiment.** Many investors use Steam concurrent players only as a sentiment proxy. A panel model that controls for release calendar, game age, seasonality and consensus can extract incremental information.
- **Coverage-bias judgment.** Steam covers PC but not console or mobile. Knowing which publishers and games are representative is part of the edge.

#### 4. Data and sources

**Checked (public documentation; not yet ingested):**

- SteamDB and Steam Charts (concurrent and peak players, followers).
- TwitchTracker and SullyGnome (hours watched, streamers).
- YouTube view counts and Reddit/community activity.
- Google Play and App Store rankings.
- Company 10-K/10-Q/20-F, earnings calls and shareholder letters.
- Circana (US game sales) and GSD (Europe).

**Paid / still to verify:**

- Sensor Tower, AppMagic or data.ai for mobile; FactSet/Bloomberg/Visible Alpha for consensus.
- Licensing and terms of service for scraping; historical coverage; API stability; whether the data is point-in-time.

#### 5. Research approach

- **Game-level panel:** daily or weekly concurrent players, hours watched, review counts and rank; map each game to a publisher and revenue weight.
- **Signal:** four-week engagement change versus a seasonal baseline or the prior year, weighted by estimated revenue contribution; or a "beat/miss" nowcast of quarterly revenue.
- **Decision:** long publishers with positive engagement surprises and high live-service revenue share; short negative surprises; or pair against the sector ETF. Event-driven extension: fade launch spikes when decay is faster than the typical pattern.
- **Tests:** panel regression of revenue surprise or guidance change on engagement surprise, controlling for release calendar, game age, platform and macro; event study around releases and earnings; out-of-sample by time and publisher.

The approach fits because engagement is a direct driver of live-service revenue and is observable at high frequency.

#### 6. Validation and risks

- **Benchmarks:** consensus revenue, sector ETF, publisher peer basket, and a simple release-date strategy.
- **Reject if:** engagement surprises have no incremental predictive power after controlling for consensus and the release calendar; profits vanish after costs; the signal is driven by one game or publisher; or coverage bias explains the result.
- **Risks:** engagement is not revenue (monetization, discounts, platform fees); scraping terms of service and legal risk; PC versus console/mobile coverage; release-calendar lumpiness; small sample; consensus may already use alternative data; shorting costs and liquidity; regime changes such as subscriptions.
- **First two-week test:** pick five live-service games and three publishers, build an eight-quarter panel, and regress revenue surprises on four-week engagement changes. Check the sign, t-statistic and out-of-sample stability before expanding the universe.

## Automation plan (Module 5 tie-in)

Each idea can be deployed as a small daily research pipeline before any capital is committed.

| Idea | Data refresh | Pipeline | Trigger | Monitoring |

| 1 | Daily local news and Defensoria; monthly MINEM; weekly satellite | Scrape, deduplicate, classify event, map to mine, score severity | New high-severity event or score above threshold | Event-mapping precision; latency versus English media |
